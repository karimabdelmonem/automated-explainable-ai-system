```python
# Updated Load Data Cell
import pandas as pd
import os

file_path = r'C:\Users\User\Desktop\XAI_System\data\raw\Loan_default.csv'
df = pd.read_csv(file_path)

# Basic Data Inspection
print("DATASET DIMENSIONS")
print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
display(df.head())
```

## Step 3: Data Quality Assessment

```python
print("DF.INFO() - Complete Overview")
df.info()

print("\nMISSING VALUES ANALYSIS")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing_Count': missing,'Missing_%': missing_pct}).sort_values('Missing_%', ascending=False)
display(missing_df[missing_df['Missing_Count'] > 0])

# Visualize missing values
if missing.sum() > 0:
    plt.figure(figsize=(10, 6))
    sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
    plt.title('Missing Values Heatmap', fontsize=14, fontweight='bold')
    plt.show()

print("\nDUPLICATE CHECK")
print(f"• Duplicate rows: {df.duplicated().sum():,}")
print(f"• Duplicate %: {(df.duplicated().sum()/len(df)*100):.3f}%")

print("\nUNIQUE VALUES PER COLUMN")
for col in df.columns:
    print(f"• {col:20s}: {df[col].nunique():6,} unique values")
```

## Step 4: Statistical Summary

```python
print("NUMERICAL FEATURES - DESCRIBE()")
display(df.describe().T)

print("\nCATEGORICAL FEATURES SUMMARY")
categorical_cols = df.select_dtypes(include=['object', 'category']).columns
if len(categorical_cols) > 0:
    for col in categorical_cols:
        print(f"\n{col.upper()}")
        display(df[col].value_counts().head(10).to_frame(name='Count'))
else:
    print("No categorical columns detected")

# Target variable analysis (assuming 'Default' or similar)
target_col = [col for col in df.columns if col.lower() in ['default', 'target', 'loan_status', 'outcome']]
if target_col:
    target_col = target_col[0]
    print(f"\nTARGET VARIABLE: '{target_col}'")
    display(df[target_col].value_counts(normalize=True).to_frame(name='Distribution %').style.format('{:.2%}'))

    # Visualize target distribution
    fig = px.pie(names=df[target_col].value_counts().index,
                 values=df[target_col].value_counts().values,
                 title=f'{target_col} Distribution',
                 color_discrete_sequence=px.colors.sequential.RdBu)
    fig.update_traces(textposition='inside', textinfo='percent+label')
    fig.show()
```

### More 3D Scatter Plots

```python
# 6. Second 3D Scatter Plot: Age, LoanAmount, and CreditScore by Default
if len(numerical_cols) >= 3 and target_col:
    x_feat, y_feat, z_feat = 'Age', 'LoanAmount', 'CreditScore'
    fig = px.scatter_3d(df.sample(min(2000, len(df)), random_state=42),
                        x=x_feat, y=y_feat, z=z_feat,
                        color=target_col,
                        title=f'3D View: {x_feat} x {y_feat} x {z_feat} by Default Status',
                        color_discrete_map={0: '#2ecc71', 1: '#e74c3c'})
    fig.update_layout(scene=dict(
        xaxis_title=x_feat, yaxis_title=y_feat, zaxis_title=z_feat))
    fig.show()
```

```python
# 7. Third 3D Scatter Plot: Income, MonthsEmployed, and InterestRate by Default
if len(numerical_cols) >= 3 and target_col:
    x_feat, y_feat, z_feat = 'Income', 'MonthsEmployed', 'InterestRate'
    fig = px.scatter_3d(df.sample(min(2000, len(df)), random_state=42),
                        x=x_feat, y=y_feat, z=z_feat,
                        color=target_col,
                        title=f'3D View: {x_feat} x {y_feat} x {z_feat} by Default Status',
                        color_discrete_map={0: '#2ecc71', 1: '#e74c3c'})
    fig.update_layout(scene=dict(
        xaxis_title=x_feat, yaxis_title=y_feat, zaxis_title=z_feat))
    fig.show()
```

### Even More 3D Scatter Plots

```python
# 8. Fourth 3D Scatter Plot: NumCreditLines, LoanTerm, and DTIRatio by Default
if len(numerical_cols) >= 3 and target_col:
    x_feat, y_feat, z_feat = 'NumCreditLines', 'LoanTerm', 'DTIRatio'
    fig = px.scatter_3d(df.sample(min(2000, len(df)), random_state=42),
                        x=x_feat, y=y_feat, z=z_feat,
                        color=target_col,
                        title=f'3D View: {x_feat} x {y_feat} x {z_feat} by Default Status',
                        color_discrete_map={0: '#2ecc71', 1: '#e74c3c'})
    fig.update_layout(scene=dict(
        xaxis_title=x_feat, yaxis_title=y_feat, zaxis_title=z_feat))
    fig.show()
```

```python
# 5. Interactive Bar Charts for Categorical Features vs. Default Rate
categorical_features_to_plot = ['Education', 'EmploymentType', 'MaritalStatus', 'LoanPurpose', 'HasMortgage', 'HasDependents', 'HasCoSigner']

for col in categorical_features_to_plot:
    if col in df.columns and target_col:
        # Calculate default rate for each category
        default_rate_by_category = df.groupby(col)[target_col].mean().reset_index()
        default_rate_by_category.columns = [col, 'Default_Rate']
        default_rate_by_category['Default_Rate_Percent'] = (default_rate_by_category['Default_Rate'] * 100).round(2)

        fig = px.bar(default_rate_by_category,
                     x=col,
                     y='Default_Rate_Percent',
                     title=f'Default Rate by {col}',
                     color=col, # Color by category
                     color_discrete_sequence=px.colors.qualitative.Pastel,
                     labels={'Default_Rate_Percent': 'Default Rate (%)'},
                     text_auto=True # Show text labels on bars
                    )
        fig.update_layout(yaxis_range=[0, default_rate_by_category['Default_Rate_Percent'].max() * 1.1]) # Adjust y-axis to fit labels
        fig.show()
```
